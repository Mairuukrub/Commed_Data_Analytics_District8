```python
hb = d[(d["report_code"].str.startswith("fdc28c")) & (d["scope"] == "unit")]

print(hb["column"].value_counts())
print(hb["row_name"].value_counts().head(10))
print(hb.groupby("fiscal_year_be").size())

#ได้เห็นว่าข้อมูลปี 69 นั้นมีการจัดไม่เหมือนรูปแบบอื่นๆ 
```

```python
pv = hb.pivot_table(index="fiscal_year_be", columns="column",
                    values="value_num", aggfunc="first")

pv = pv[["target", "result", "F3", "target1", "result1", "F6"]]
pv["pct1_คำนวณ"] = pv["result"] / pv["target"] * 100
pv["pct2_คำนวณ"] = pv["result1"] / pv["target1"] * 100

print(pv.round(2).to_string())

#ทำ pivot table ตรวจสอบข้อมูลจริง ของ HbA1c 
#target = จำนวนผู้ป่วย (B1) , Result = ได้รับการตรวจ HbA1c(A1) , F3 คือ ร้อยละการตรวจ  (เป็น ผู้ป่วยที่อยู่ในเขตรับผิดชอบ Typearea 1,3) | target1 เหมือนกันแค่เป็นแบบ ผู้ป่วยที่มารับบริการของหน่วยบริการจากแฟ้ม ChronicFU
```

```python
import numpy as np

def wilson(x, n, z=1.96):
    p = x / n
    denom = 1 + z**2 / n
    center = (p + z**2 / (2 * n)) / denom
    half = z * np.sqrt(p * (1 - p) / n + z**2 / (4 * n**2)) / denom
    return center - half, center + half

pv["lo"], pv["hi"] = wilson(pv["result"], pv["target"])
pv["lo"] *= 100
pv["hi"] *= 100

print(pv[["target", "result", "pct1_คำนวณ", "lo", "hi"]].round(1).to_string())
```

```python
import matplotlib as plt
import numpy as np

years = pv.index.to_numpy()
x = np.arange(len(years))
w = 0.38

fig, ax = plt.subplots(figsize=(8, 4.5))

b1 = ax.bar(x - w/2, pv["target"], w, label="ผู้ป่วยในเขต (target)")
b2 = ax.bar(x + w/2, pv["result"], w, label="ได้รับการตรวจ HbA1c (result)")
ax.bar_label(b1, fmt="%d", padding=2)
ax.bar_label(b2, fmt="%d", padding=2)

ax.set_xticks(x)
ax.set_xticklabels(years)
ax.set_ylim(0, pv["target"].max() * 1.25)
ax.set_xlabel("ปีงบประมาณ (พ.ศ.)")
ax.set_ylabel("จำนวนคน")

ax2 = ax.twinx()
ax2.plot(x, pv["pct1_คำนวณ"], color="black", marker="o", label="ร้อยละ")
ax2.set_ylim(-60, 100)
ax2.set_ylabel("ร้อยละที่ได้รับการตรวจ")

# ถ้ารู้เกณฑ์ที่เชื่อถือได้ ใส่ตรงนี้ (ตัวอย่าง: goal = 80) ไม่รู้ก็ปล่อยเป็น None
goal = None
if goal is not None:
    ax2.axhline(goal, linestyle="--", color="gray", label=f"เกณฑ์ {goal}%")

h1, l1 = ax.get_legend_handles_labels()
h2, l2 = ax2.get_legend_handles_labels()
ax.legend(h1 + h2, l1 + l2, loc="upper left")

ax.set_title("ผู้ป่วยเบาหวานที่ได้รับการตรวจ HbA1c อย่างน้อย 1 ครั้ง/ปี")
plt.tight_layout()
plt.show()
```
